```python
import requests
import base64

# Google Cloud Vision API 엔드포인트
url = "https://vision.googleapis.com/v1/images:annotate"

# API 키 (문자열 형식)
api_key = "AIzaSyCEv0_bOGZu2g7y9vAN-tibfGOMaJWO__c"

# 이미지 파일 경로
image_path = r'D:\2024신한해커톤\영수증 처리 프로젝트\test\example.jpg'

# 이미지 파일을 읽어서 Base64로 인코딩
with open(image_path, "rb") as image_file:
    encoded_image = base64.b64encode(image_file.read()).decode()

# 요청 페이로드 생성
payload = {
    "requests": [
        {
            "image": {
                "content": encoded_image
            },
            "features": [
                {
                    "type": "TEXT_DETECTION"
                }
            ]
        }
    ]
}

# API 요청 보내기
response = requests.post(url, json=payload, params={"key": api_key})

# 결과 출력
if response.status_code == 200:
    result = response.json()
    for annotation in result['responses'][0]['textAnnotations']:
        print(annotation['description'])
else:
    print(f"Error: {response.status_code}")
    print(response.text)
```

```python
import openai

# OpenAI API 키 설정
openai.api_key = ''

def process_receipt_text(receipt_text):
    # GPT-4 모델에 입력할 프롬프트
    prompt = f"""
    다음은 영수증에서 OCR을 통해 추출된 텍스트입니다:

    텍스트: "{receipt_text}"

    1. 이 텍스트에서 결제 시간을 찾아서 'YYYY-MM-DD HH:MM' 형식으로 반환하세요.
    2. 결제 시 텀블러를 사용해 절약했는지 여부를 확인하세요. "텀블러 할인" 또는 "할인 없음"으로만 대답하세요.

    답변 형식:
    결제 시간: YYYY-MM-DD HH:MM
    텀블러 사용: 텀블러 할인 / 할인 없음
    """

    response = openai.Completion.create(
        engine="text-davinci-003",  # GPT-3.5 기준, 혹은 "gpt-4"로 변경
        prompt=prompt,
        max_tokens=100,
        n=1,
        stop=None,
        temperature=0.2,
    )

    return response.choices[0].text.strip()

# 예시 영수증 텍스트
receipt_text = "2024-08-22 14:53 결제 ... 텀블러 할인 300원 ..."

# 함수 호출
result = process_receipt_text(receipt_text)
print(result)
```
