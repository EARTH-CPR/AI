```python
import os
import tensorflow as tf
from tensorflow.keras.preprocessing.image import save_img

# 경로 설정
input_dir = r'D:\2024신한해커톤\음식 섭취 추정 프로젝트\Data/Before'  # 원본 이미지들이 들어있는 폴더 경로
output_dir = r'D:\2024신한해커톤\음식 섭취 추정 프로젝트\Data/Before/augmented'  # 증식된 이미지를 저장할 폴더 경로

# 폴더가 없다면 생성
if not os.path.exists(output_dir):
    os.makedirs(output_dir)

# 경로 내의 모든 이미지를 증식
for filename in os.listdir(input_dir):
    if filename.endswith('.jpg') or filename.endswith('.png'):  # 원하는 이미지 포맷에 맞게 확장자 설정
        # 이미지 파일 경로
        image_path = os.path.join(input_dir, filename)

        # 이미지 로드
        image = tf.keras.preprocessing.image.load_img(image_path)
        x = tf.keras.preprocessing.image.img_to_array(image)
        x = tf.expand_dims(x, 0)  # 이미지 배열 차원 확장

        # 이미지 증식
        for i in range(10):  # 10배로 증식
            augmented_image = tf.image.random_flip_left_right(x)
            augmented_image = tf.image.random_flip_up_down(augmented_image)
            augmented_image = tf.image.random_brightness(augmented_image, max_delta=0.2)
            augmented_image = tf.image.random_contrast(augmented_image, lower=0.5, upper=1.5)
            
            # 차원 축소
            augmented_image = tf.squeeze(augmented_image, axis=0) 

            # 파일 저장
            new_filename = f"{os.path.splitext(filename)[0]}_aug_{i}.jpg"
            save_path = os.path.join(output_dir, new_filename)
            save_img(save_path, augmented_image)
```
