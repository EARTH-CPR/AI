```python
from tensorflow.keras.models import load_model, Model

# 모델 로드
model = load_model(r'C:\2024신한해커톤\운동 인식 프로젝트\model/worksearch_model.h5')
```

```python
import matplotlib.pyplot as plt
from tensorflow.keras.preprocessing import image
import os
import numpy as np

from matplotlib import font_manager, rc

# 한글 폰트 설정 (예: 'Malgun Gothic' 또는 'NanumGothic')
plt.rcParams['font.family'] = 'Malgun Gothic'  # Windows의 경우

# 이미지 폴더 경로
img_folder = r'C:\2024신한해커톤\운동 인식 프로젝트\Data\testImage\passImg'
img_files = os.listdir(img_folder)

# 최대 36개의 이미지만 선택
img_files = img_files[:36]

# 그림 크기 설정
fig, axes = plt.subplots(6, 6, figsize=(12, 12))

# 이미지 파일들을 불러와서 6x6 그리드에 시각화
for i, img_file in enumerate(img_files):
    img_path = os.path.join(img_folder, img_file)
    img = image.load_img(img_path, target_size=(150, 150))
    
    # 이미지를 배열로 변환 및 전처리
    img_array = image.img_to_array(img)
    img_array = np.expand_dims(img_array, axis=0)
    img_array = img_array / 255.0
    
    # 예측
    preds = model.predict(img_array)
    pred_value = preds[0][0]  # 예측된 확률값 (이진 분류인 경우)

    # 현재 subplot을 지정
    ax = axes[i // 6, i % 6]
    
    # 이미지 표시
    ax.imshow(img)
    ax.axis('off')
    ax.set_title(f"예측값: {pred_value:.4f}")

# 여백을 조정하여 보기 좋게 만듭니다.
plt.tight_layout()

save_path = r'C:\2024신한해커톤\운동 인식 프로젝트\result/output_passImg.png'
plt.savefig(save_path, dpi=300, bbox_inches='tight')

plt.show()
```
