```python
import json
import os
import csv

# JSON 파일이 있는 디렉토리 경로
json_dir = r'C:\2024신한해커톤\운동 인식 프로젝트\Data\013.피트니스자세_sample\라벨링데이터'
# 결과를 저장할 CSV 파일 경로
csv_file = 'output.csv'

# CSV 파일을 생성하고, 헤더를 작성합니다.
with open(csv_file, mode='w', newline='', encoding='utf-8') as file:
    writer = csv.writer(file)
    writer.writerow(['active', 'img_key'])

    # JSON 디렉토리의 모든 파일을 읽습니다.
    for filename in os.listdir(json_dir):
        if filename.endswith('.json'):
            file_path = os.path.join(json_dir, filename)
            
            with open(file_path, 'r', encoding='utf-8') as f:
                try:
                    data = json.load(f)
                    
                    # JSON에서 frames 데이터를 반복합니다.
                    for frame in data['frames']:
                        for view_key, view_data in frame.items():
                            # 'active'와 'img_key' 데이터를 추출합니다.
                            active = view_data.get('active', 'No')
                            img_key = view_data.get('img_key', '')
                            
                            # 데이터를 CSV 파일에 기록합니다.
                            writer.writerow([active, img_key])
                except json.JSONDecodeError as e:
                    print(f"Error decoding JSON from file {filename}: {e}")
                except UnicodeDecodeError as e:
                    print(f"Error decoding file {filename}: {e}")
```

```python
import pandas as pd

# CSV 파일 경로
# csv_file = 'path_to_your_csv_file.csv'

# CSV 파일 읽기
df = pd.read_csv(csv_file)

# img_key 열에서 마지막 파일명을 추출
df['img_key'] = df['img_key'].apply(lambda x: x.split('/')[-1])

# 변환된 데이터를 새로운 CSV 파일로 저장
df.to_csv('output2.csv', index=False)

print("변환된 CSV 파일이 'output.csv'로 저장되었습니다.")
```
